Explicación de slicing
La técnica de "slicing" en Python se utiliza para extraer una porción o subcadena de un string. Permite acceder a partes específicas del string utilizando índices o rangos.

El slicing se realiza utilizando la sintaxis __[inicio:fin]__, donde:

- inicio es el índice de inicio del subcadena (incluido).
- fin es el índice de fin del subcadena (excluido).

**5) Índices en cadenas**

**6) Extraer partes de una dirección**

Dada la variable `direccion` con el valor `"Calle Falsa, 123, Springfield"`:

1. Extrae e imprime solo el nombre de la calle.
2. Extrae e imprime solo el número de la calle.
3. Extrae e imprime solo la ciudad.

📌 **Nota**: Más adelante veremos como separar esto de una forma más automatizada, separando la información por comas por ejemplo, pero de momento se realizará utilizando slicing normal.

````python
# Creamos la variable tipo (String)
direccion = "Calle Falsa, 123, Springfield"

# Extraemos e imprimimos solo el nombre de la calle
print(direccion[0:11]) # El resultado es: Calle Falsa

# Extraemos e imprimimos el número de la calle
print(direccion[13:16]) # El resultado es: 123

# Extraemos e imprimimos solo la ciudad
print(direccion[18:99]) # El resultado es: Springfield "pongo 99 para no tardar
````

### **7) Crear un identificador único con slicing y concatenación**

En este ejercicio, crearemos un identificador único utilizando **slicing** y **concatenación**.

1. Define tres variables:
   - `nombre`: Tu nombre (por ejemplo, `"Cristian"`).
   - `apellido`: Tu apellido (por ejemplo, `"Rodríguez"`).
   - `año_nacimiento`: Tu año de nacimiento (por ejemplo, `"1989"`).
       - -> IMPORTANTE. El año esta también como str, vamos a trabajar con todo strings de momento

2. Extrae las siguientes partes de cada variable usando slicing:
   - **Las tres primeras letras** del `nombre`.
   - **Las tres primeras letras** del `apellido`.
   - **Los dos últimos dígitos** del `año_nacimiento`.

3. Concatena estos valores para formar un identificador único con el siguiente formato:

    __CriRod89__

4. Imprime el identificador resultante.

📌 **Nota**: Puedes usar tus propios datos en lugar de los del ejemplo.


````python
# Definimos las 3 variables tipo strings
nombre = "Yasin"
apellido = "Benkssouba"
año_nacimiento = "1999"
# Extraemos las partes solicitadas de cada variable usando Slincing
nombre_parte = nombre[0:3]
apellido_parte = apellido[0:3]
año_parte = año_nacimiento[-2:]
# concatenamos los valores para formar el "identificador"
identificador = nombre_parte + apellido_parte + año_parte
# Imprimo el resultado
print(identificador)
````
